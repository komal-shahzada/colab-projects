```python
# Step 1: Install Gradio (if not already)
!pip install -q gradio

# Step 2: Import libraries
from google import genai
import gradio as gr

# Step 3: Configure Gemini API
client = genai.Client(api_key="AIzaSyDra9-p3FHAcR9jYOfZW9cD3q-MTSX0C70")  # Use your actual API key

# Step 4: Function to talk to Gemini
def chat_with_gemini(user_input):
    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=f"You are a news and weather assistant. Reply with news headlines or current weather info.\nUser: {user_input}"
        )
        return response.text
    except Exception as e:
        return f"❌ Error: {str(e)}"

# Step 5: Build Gradio UI with custom HTML/CSS styling
with gr.Blocks(css="""
body {
    background: linear-gradient(135deg, #fce4ec, #f3e5f5);
    font-family: 'Segoe UI', sans-serif;
}
h1 {
    color: #d63384;
}
p {
    color: #FFFFFF;
    font-size: 16px;
}
.gr-button-primary {
    background-color: #d63384 !important;
    color: white !important;
    border: none;
}
.gr-button-secondary {
    background-color: #6a1b9a !important;
    color: white !important;
    border: none;
}
textarea, input {
    border: 2px solid #f06292 !important;
    background-color: #fce46c !important;
    color: #4a144c !important;
}
.gr-chatbot {
    background-color: #f8bbd0 !important;
}
""") as interface:

    gr.Markdown(
        """
        <h1 style='text-align: center;'>🌸 News & Weather Chatbot</h1>
        <p style='text-align: center;'>Ask about today's headlines, global events, or the weather in any city!</p>
        """
    )

    chatbot = gr.Chatbot(height=400)
    user_input = gr.Textbox(placeholder="Type a news or weather question...", label="Your Question")
    send_btn = gr.Button("Send", variant="primary")
    reset_btn = gr.Button("Reset Chat", variant="secondary")

    def respond(message, history):
        response = chat_with_gemini(message)
        history = history or []
        history.append((message, response))
        return history, ""

    def clear_chat():
        return [], ""

    send_btn.click(respond, inputs=[user_input, chatbot], outputs=[chatbot, user_input])
    reset_btn.click(clear_chat, outputs=[chatbot, user_input])

interface.launch()
```
