```python
# Step 1: Install MoviePy
!pip install moviepy

# Step 2: Upload your assets
from google.colab import files
print("📤 Please upload 'email8_Screenshot_1', 'email8_animation', and 'email8_Screenshot_2'")
uploaded = files.upload()

# Step 3: Create scrolling video
from moviepy.editor import ImageClip, VideoFileClip, CompositeVideoClip, ColorClip
import os

# ==== SETTINGS ====
VIDEO_WIDTH = 1080
VIDEO_HEIGHT = 1920
CONTENT_WIDTH = 800
SCROLL_SPEED = 243.1
FPS = 60
TOP_SPACING = 200
BOTTOM_SPACING = 150

print("\n🎬 Creating video...")

# ✅ Find uploaded files
uploaded_files = list(uploaded.keys())
print(f"📂 Uploaded files: {uploaded_files}")

# Find correct file names
img1_file = [f for f in uploaded_files if 'email8_Screenshot_1' in f][0]
gif_file = [f for f in uploaded_files if 'email8_animation' in f][0]
img2_file = [f for f in uploaded_files if 'email8_Screenshot_2' in f][0]

print(f"\n✓ Using files:")
print(f"  • {img1_file}")
print(f"  • {gif_file}")
print(f"  • {img2_file}")

# ✅ Load media
img1 = ImageClip(img1_file).resize(width=CONTENT_WIDTH)
gif  = VideoFileClip(gif_file).resize(width=CONTENT_WIDTH).set_duration(0)
img2 = ImageClip(img2_file).resize(width=CONTENT_WIDTH)

print(f"\n✓ img1: {CONTENT_WIDTH}x{img1.h}px")
print(f"✓ gif:  {CONTENT_WIDTH}x{gif.h}px")
print(f"✓ img2: {CONTENT_WIDTH}x{img2.h}px")

# Total canvas height
total_height = TOP_SPACING + img1.h + gif.h + img2.h + BOTTOM_SPACING
scroll_distance = total_height - VIDEO_HEIGHT
print(f"\n📏 Total canvas height: {total_height}px")
print(f"📏 Scroll distance: {scroll_distance}px")

# Duration based on scroll speed
DURATION = scroll_distance / SCROLL_SPEED
print(f"🕒 Duration: {DURATION:.2f} seconds")

# Set GIF duration to match scrolling video
gif = gif.set_duration(DURATION)

# Create white background
background = ColorClip(size=(VIDEO_WIDTH, total_height), color=(255, 255, 255)).set_duration(DURATION)

# Stack elements
y_position = TOP_SPACING
clips = [background]
clips.append(img1.set_position(("center", y_position)))
y_position += img1.h
clips.append(gif.set_position(("center", y_position)))
y_position += gif.h
clips.append(img2.set_position(("center", y_position)))

stacked = CompositeVideoClip(clips, size=(VIDEO_WIDTH, total_height)).set_duration(DURATION)

# Scroll effect
def scroll_effect(get_frame, t):
    y_offset = int((scroll_distance / DURATION) * t)
    frame = get_frame(t)
    return frame[y_offset:y_offset + VIDEO_HEIGHT, :, :]

scrolling_clip = stacked.fl(scroll_effect).set_duration(DURATION)

print("🎥 Rendering video... (this may take a few minutes)")
scrolling_clip.write_videofile(
    "scroll_video.mp4",
    fps=FPS,
    codec="libx264",
    bitrate="5000k",
    audio=False,
    preset='medium'
)

print("\n✅ Video created successfully!")
files.download("scroll_video.mp4")
```
